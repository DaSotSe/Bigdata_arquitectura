# Intro a AWS

## Visión General AWS. Comparando con GCP

*Amazon Web Services (AWS)* es la plataforma de computación en la nube más utilizada a nivel mundial, ofreciendo más de 200 servicios en distintas áreas, como cómputo, almacenamiento, bases de datos, redes, machine learning, seguridad y analítica. Fue lanzada en 2006 y actualmente domina el mercado cloud con una cuota de más del 30%, seguido por Azure y GCP.

image.png

| Característica        | AWS                                                   | GCP                                                        |
|----------------------|-----------------------------------------------------|----------------------------------------------------------|
| Cantidad de Servicios | +200 servicios, el ecosistema más grande.           | Menos servicios, pero con fuerte integración en datos y AI. |
| Estrategia de Mercado | Prioriza flexibilidad y variedad de opciones.        | Optimizado para análisis de datos, inteligencia artificial y machine learning. |
| Facilidad de Uso      | Más complejo de aprender por su amplitud.            | UI más sencilla y herramientas bien integradas.           |
| Facturación          | Más opciones de precios (spot, reservadas, on-demand). | Más sencillo de entender y generalmente más barato en analítica. |
| Networking          | AWS VPC permite redes más configurables.               | GCP ofrece redes más eficientes con menor latencia.       |

## 2- Servicios principales y sus equivalentes en GCP

| **Servicio**      | **AWS**                                  | **GCP**                      |
|-------------------|------------------------------------------|------------------------------|
| **Cómputo**       | EC2 (Elastic Compute Cloud)              | Compute Engine               |
| **Almacenamiento**| S3 (Simple Storage Service)              | Cloud Storage                |
| **Bases de Datos**| RDS (Relational Database Service), DynamoDB (Amazon DynamoDB) | Cloud SQL, Bigtable          |
| **Big Data**      | EMR (Elastic MapReduce)                  | Dataproc                     |
| **Data Warehouse**| Redshift (Amazon Redshift)               | BigQuery                     |
| **Serverless**    | Lambda (AWS Lambda)                     | Cloud Functions              |
| **Mensajería**    | SQS (Simple Queue Service), SNS (Simple Notification Service), EventBridge (Amazon EventBridge) | Pub/Sub                      |


Si bien AWS tiene más servicios, *GCP es más sencillo de utilizar* y además está más orientado y optimizado hacia el propósito de *datos e IA*.

En cuanto a los costos, AWS es más flexible, pero GCP suele ser más barato para la analítica de datos.
